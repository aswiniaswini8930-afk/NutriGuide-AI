```python
from groq import Groq
from google.colab import userdata

groq_client = Groq(
    api_key=userdata.get("GROQ_API_KEY")
)

MODEL = "openai/gpt-oss-120b"

print("Python Teaching Agent connected to Groq!")
```

```python
TEACHER_SYSTEM_PROMPT = """
You are a friendly and patient Python programming tutor.

Your student is a complete beginner who has never
learned programming before.

Your teaching responsibilities:

1. Explain Python concepts using simple language.
2. Avoid complicated technical terminology.
3. Use relatable real-world analogies wherever possible.
4. Provide simple and correct Python code examples.
5. Explain code examples line by line.
6. Guide students through problems instead of
   immediately giving all the answers.
7. Ask small practice questions to check understanding.
8. Adapt explanations to the student's understanding.
9. Be encouraging and never make students feel
   uncomfortable for asking basic questions.

You can teach:
- Variables
- Data types
- Conditional statements
- Loops
- Functions

Always teach step by step.
"""

print("Python Teaching Agent system prompt created!")
```

```python
def python_teaching_agent(user_question):

    response = groq_client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "system",
                "content": TEACHER_SYSTEM_PROMPT
            },
            {
                "role": "user",
                "content": user_question
            }
        ],
        temperature=0.7
    )

    teaching_response = response.choices[0].message.content

    return teaching_response


print("Python Teaching Agent is ready!")
```

```python
student_question = "What is a Python variable?"

teaching_response = python_teaching_agent(student_question)

print("STUDENT QUESTION:")
print(student_question)

print("\nPYTHON TEACHING AGENT:")
print(teaching_response)
```

```python
JUDGE_SYSTEM_PROMPT = """
You are an expert evaluator of Python teaching quality.

Your task is to evaluate responses generated by a
Python Teaching Agent for complete beginners.

Evaluate the response using these five criteria.

1. conceptual_accuracy:
   Is the Python information technically correct?
   Score: 0 to 5.

2. clarity:
   Is the explanation clear and easy to understand?
   Score: 0 to 5.

3. beginner_friendliness:
   Is the explanation suitable for someone with
   absolutely no programming knowledge?
   Score: 0 to 5.

4. example_quality:
   Are the examples relevant, correct, and helpful?
   Score: 0 to 5.

5. teaching_approach:
   Does the response guide the learner, explain
   step by step, and encourage practice?
   Score: 0 to 5.

Scoring rules:
- 0 means completely inadequate.
- 1 means very poor.
- 2 means needs improvement.
- 3 means satisfactory.
- 4 means very good.
- 5 means excellent.

Calculate total_score as the sum of all five scores.

Return ONLY valid JSON with this exact structure:

{
    "conceptual_accuracy": 0,
    "clarity": 0,
    "beginner_friendliness": 0,
    "example_quality": 0,
    "teaching_approach": 0,
    "total_score": 0,
    "feedback": "Brief evaluation of the teaching response.",
    "suggestions": "Specific suggestions for improvement."
}

All scores must be integers between 0 and 5.
The total score must be between 0 and 25.
Do not give scores based on politeness alone.
Evaluate the actual quality and correctness of teaching.
"""

print("LLM Judge system prompt created!")
```

```python
import json

def evaluate_teaching_response(student_question, teaching_response):

    response = groq_client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "system",
                "content": JUDGE_SYSTEM_PROMPT
            },
            {
                "role": "user",
                "content": f"""
                Student Question:
                {student_question}

                Teaching Agent Response:
                {teaching_response}

                Evaluate this teaching response.
                """
            }
        ],
        temperature=0.2,
        response_format={"type": "json_object"}
    )

    evaluation = response.choices[0].message.content

    return json.loads(evaluation)


print("LLM Judge is ready!")
```

```python
evaluation = evaluate_teaching_response(
    student_question,
    teaching_response
)

print(json.dumps(evaluation, indent=4))
```

```python
def python_learning_system(student_question):

    # Agent 1: Generate teaching response
    teaching_response = python_teaching_agent(student_question)

    # Agent 2: Evaluate teaching response
    evaluation = evaluate_teaching_response(
        student_question,
        teaching_response
    )

    # Display results
    print("=" * 50)
    print("PYTHON TEACHING AGENT")
    print("=" * 50)

    print(teaching_response)

    print("\n" + "=" * 50)
    print("LLM JUDGE EVALUATION")
    print("=" * 50)

    print(json.dumps(evaluation, indent=4))

    return teaching_response, evaluation


print("Python Teaching and Evaluation System is ready!")
```

```python
teaching_response, evaluation = python_learning_system(
    "I am completely new to programming. What are Python variables?"
)
```

```python
teaching_response_2, evaluation_2 = python_learning_system(
    "I am a complete beginner. Explain Python if-else statements with a simple real-world example."
)
```

```python
teaching_response_3, evaluation_3 = python_learning_system(
    "I am completely new to programming. What are Python for loops? Explain with a simple example."
)
```
