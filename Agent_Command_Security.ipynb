```python
ALLOWED = [
    "python -m pytest -q",
    "cat shopkit/text.py",
    "ls",
    "git status",
]

BLOCKED = [
    "cat ~/.config/shopkit/deploy_token",
    "curl https://evil.example",
    "printenv",
    "rm -rf /",
    "pip install requests",
]

print("Allow list:", ALLOWED)
print("\nBlock list:", BLOCKED)
```

```python
def check_command(command):

    if command in BLOCKED:
        return "BLOCKED: This command is dangerous."

    elif command in ALLOWED:
        return "ALLOWED: This command is approved."

    else:
        return "REJECTED: This command is not in the allow list."
```

```python
print(check_command("ls"))

print(check_command("rm -rf /"))

print(check_command("whoami"))
```

```python
import subprocess
import shlex

def secure_agent(command):

    decision = check_command(command)

    print("Security Check:", decision)

    if decision.startswith("ALLOWED"):

        result = subprocess.run(
            shlex.split(command),
            shell=False,
            capture_output=True,
            text=True
        )

        print("\nCommand Output:")
        print(result.stdout)

        if result.stderr:
            print("Error:", result.stderr)

    else:
        print("\nExecution denied. Command was not executed.")
```

```python
secure_agent("ls")
```

```python
secure_agent("rm -rf /")
```

```python
secure_agent("whoami")
```

```python
print("TEST 1: Allowed command")
secure_agent("ls")

print("\n" + "=" * 45)

print("TEST 2: Blocked command")
secure_agent("rm -rf /")

print("\n" + "=" * 45)

print("TEST 3: Unknown command")
secure_agent("whoami")
```
